```python
%reset
import numpy as np
import matplotlib.pyplot as plt
import skimage.io as io
from tensorflow import keras
from PIL import Image
import numpy as np
import mtcnn

from keras_vggface.vggface import VGGFace
from keras_vggface.utils import preprocess_input
from keras_vggface.utils import decode_predictions
```

#FUNZIONI

```python
def estrazione_volto (filename):

  img = io.imread(filename)

  detector = mtcnn.MTCNN()
  results = detector.detect_faces(img)

  #ESTRAZIONE DEL BOX CON IL VOLTO
  x1, y1, width, height = results[0]['box']
  x2, y2 = x1 + width, y1 + height

  # extract the face
  face = img[y1:y2, x1:x2]

  image = Image.fromarray(face)
  image = image.resize((224, 224))
  face_array = np.asarray(image) #aggiunge 1 alla prima dimensione

  return face_array
```

```python
def preparazione_input (filename):

    pixels = estrazione_volto(filename)
    pixels = pixels.astype('float32')
    samples = np.expand_dims(pixels, axis=0)
    samples = preprocess_input(samples, version=2)

    return samples
```

```python
def visualizzazione_immagine (img):
    plt.figure( figsize = (15,9) )
    plt.imshow(img)
```

#MODELLO

```python
#CRERAZIONE MODELLO
from keras_vggface.vggface import VGGFace
model = VGGFace(model='resnet50')

print('Inputs: %s' % model.inputs)
print('Outputs: %s' % model.outputs)
model.summary()
```

```python
# load image from file

filename = '/content/Katy-Perry--1024x576.jpg'
samples = preparazione_input(filename)
```

```python
# perform prediction
yhat = model.predict(samples)
# convert prediction into names
results = decode_predictions(yhat)
# display most likely results

for result in results[0]:
 print('%s: %.3f%%' % (result[0], result[1]*100))
```
